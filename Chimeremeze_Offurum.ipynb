### Task 6: Imputation

#### Replace missing values with Median. LotFrontage as chosen column

```python
# Number of null values present
df['LotFrontage'].isnull().sum()
```

```python
# null value present at index 7
df.LotFrontage.head(10)
```

```python
# index 7 NaN has been replaced with median
df['LotFrontage'].fillna(df['LotFrontage'].median()).head(10)
```

#### Imputation of missing values with Linear interpolation

```python
# Use linear interpolate instead of median on LotFrontage column
df['LotFrontage'].interpolate(method = 'linear').head(10)   # the NaN at index 7 is replaced with 63.0
```

#### Imputation of missing values in categorical column

```python
# Lets replace missing values in MiscFeature with the mode
df['MiscFeature'].head(10)
```

```python
df['MiscFeature'].unique()
```

```python
# WHat is the mode?
df['MiscFeature'].mode()
```

```python
# fillna with mode (Shed)
df.MiscFeature.fillna('Shed')
```

#### Task 7: Aggregation

```python
# Groupby Neighborhood then find mean of SalePrice
df.groupby('Neighborhood')['SalePrice'].mean()
```

```python
# Groupby Neighborhood then find max of GrLivArea
df.groupby('Neighborhood')['GrLivArea'].max()
```

```python
# Groupby Neighborhood then sort by avg SalePrice in descending order
df.groupby('Neighborhood')['SalePrice'].mean().sort_values(ascending=False)
```
