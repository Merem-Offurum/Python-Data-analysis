```python
#load the dataset
import pandas as pd

path = r"C:\Users\User\Documents\Python datasets\ds_salaries_dirty.xlsx"
df = pd.read_excel(path)
```

### Task 1: Clean and Validate the Salary Data

```python
# Examining the dataset
df.head()
```

```python
df.shape
```

```python
# Getting info about null values and data types
df.info()
```

#### Work_year

```python
# Checking if there are odd/bizarre values in work_year column. None found.
df.work_year.unique()
```

#### Experience_level

```python
df.experience_level.value_counts()
```

```python
# Replace null values
df['experience_level'].fillna(value='EN', inplace=True)
```

#### Employment_type

```python
df['employment_type'].fillna(value='FT', inplace=True)
```

#### Job_title

```python
df.job_title.value_counts()
```

```python
# remove entry with less than 3 characters, then reindex
df = df[~(df['job_title'] == 'DS')]
df.reset_index(inplace = True)
```

```python
# drop the indeex column that was added during reindexing
df.drop(columns = 'index', inplace = True)
```

#### Salary_in_usd

```python
#replace salary_in_usd values below 10,000 or above 1,000,000 with median.
median_salary = df.salary_in_usd.median()
df.loc[df['salary_in_usd'] < 10000, 'salary_in_usd'] = median_salary
df.loc[df['salary_in_usd'] > 1000000, 'salary_in_usd'] = median_salary
```

```python
# It appears that there are 4 null values in salary_in_usd. Replace with median below. Median is preferred to mean due to possible outliers.
# Also, these columns are not being dropped because they are only 4 and so replacing them would cause little or no bias.
df['salary_in_usd'] = df['salary_in_usd'].fillna(median_salary)
```

#### Company_location

```python
#checking for invalid entries
df.company_location.value_counts()
```

```python
df.company_location.unique()
```

```python
# replace missing values in company_location with 'US'
df['company_location'] = df['company_location'].fillna(value = 'US')
```

```python
# replace the invalid entry in company_location with 'US'
df.loc[df['company_location'] == '154019.0', 'company_location'] = 'US'
```

#### Company_size

```python
# checking for invalid entries
df.company_size.value_counts()
```

```python
df.company_size.unique()
```

```python
# replace invalid entries with 'M'
df.loc[df['company_size'] == 'US', 'company_size'] = 'M'
df.loc[df['company_size'] == 'XL', 'company_size'] = 'M'
```

```python
# My Output dataframe
clean_salaries = df.copy()
```

### Task 2: Salary Trends by Experience

```python
# group data by experience level and employment type. Then, find the average salary. All of this is then made into a dataframe.
salary_trends = pd.DataFrame(df.groupby(['experience_level', 'employment_type'])['salary_in_usd'].mean())
```

```python
salary_trends
```

```python
# find average salary of experience levels to get a clearer understanding
df.groupby('experience_level')['salary_in_usd'].mean().round()
```

```python
# find average salary of employment types to geta clearer understanding
df.groupby('employment_type')['salary_in_usd'].mean().round()
```

#### Among the various experience levels, the highest earners are EX, followed by EN and SE, while MI earn the least.
#### Across experience levels, the highest earning employment types tend to be PT and FL, followed by FT while CT earns the least.

### Task 3: Remote Work Analysis

```python
# First exclude remote_ratio of 50, then groupby remote_ratio, aggregate salary into mean. Finally, wrap everything into a dataframe and assign to a 
# variable.
remote_salary_comparison = pd.DataFrame(df[~(df['remote_ratio'] == 50)].groupby('remote_ratio')['salary_in_usd'].mean().rename(
    index = {100 : 'Remote', 0 : 'On-site'}))
```

```python
remote_salary_comparison
```

```python
# lets check how 50% on-site work compares to the other two groups
df.groupby('remote_ratio')['salary_in_usd'].mean()
```

#### Remote workers earn less on average than on-site workers. We also see that those who work 50% on-site earn more than those 
#### completely remote, but less than those who work 100% on-site.
