### Part 2: Refactor Using Functions

```python
# 1.
student_dict_list = list()

def collect_data():
    number_of_students = int(input("Enter number of students: "))
    for i in range(0, number_of_students):
        student_name = input("Enter student's name: ")
        math_score = int(input(f"Enter {student_name}'s math score: "))
        english_score = int(input(f"Enter {student_name}'s english score: "))
        science_score = int(input(f"Enter {student_name}'s science score: "))
        arts_score = int(input(f"Enter {student_name}'s arts score: "))
        student_dict_list.append({'name': student_name, 'scores': [math_score, english_score, science_score, arts_score]})
    return student_dict_list

collect_data()
```

```python
# 2.
def calculate_average(scores):
    return sum(scores)/len(scores)

calculate_average([50,55,60,65,10,20,30,40,90,80,95,89])
```

```python
# 2. Alternative method

def calculate_average_2(student_dict_list):
    for i in range(0,len(student_dict_list)):
        student_avg = sum(student_dict_list[i]['scores'])/len(student_dict_list[i]['scores'])
        print(student_avg)

calculate_average_2(student_dict_list)
```

```python
# 3. 
def assign_grade(avg):
    if avg >= 80:
       grade = 'A' 
    elif avg >= 70:
        grade = 'B'
    elif avg >= 60:
        grade = 'C'
    elif avg >= 50:
        grade = 'D'
    else:
        grade = 'F'
    return (grade)

assign_grade(88.5)
```

```python
# 4.
def generate_remark(grade):
    if grade == 'A':
        comment = "Your grade is A, excellent."
    elif grade == 'B':
        comment = "Your grade is B, very good."
    elif grade == 'C':
        comment = "Your grade is C, good."
    elif grade == 'D':
        comment = "Your grade is D, fair - needs improvement."
    else:
        comment = "Your grade is F, poor - work harder."
    return comment

generate_remark('A')
```

```python
# 5. 
import pandas as pd

def summarize_students(data):
    name = []
    avg = []
    grade = []
    remark = []
    for dict in data:
        name.append(dict['name'])
        average = sum(dict['scores'])/len(dict['scores'])
        avg.append(average) 
        stud_grd = assign_grade(average) 
        grade.append(stud_grd)
        rmrk = generate_remark(stud_grd)
        remark.append(rmrk)
    student_dict = {'name': name, 'avg': avg, 'grade': grade, 'remark': remark}
    student_tbl = pd.DataFrame(student_dict)
    return student_tbl

student_table = summarize_students([{'name': 'Jack', 'scores': [50, 55, 60, 55]}, 
                    {'name': 'Amanda', 'scores': [10, 20, 30, 40]}, 
                    {'name': 'Tom', 'scores': [90, 80, 95, 89]}])
student_table
```

### Part 3: Add Descriptive Stats

```python
# Overall class average
def class_avg():
    class_scores = []
    class_length = []
    for dict in student_dict_list:
        class_scores.append(sum(dict['scores']))
        class_length.append(len(dict['scores']))
    return sum(class_scores)/sum(class_length)

class_avg()
```

```python
# Highest and lowest scoring student
import pandas as pd

def highest():
    student_table_sorted_descending = student_table.sort_values(by='avg', ascending =False, ignore_index = True)
    high = student_table_sorted_descending.loc[0,'name']
    high_score = student_table_sorted_descending.loc[0,'avg']
    print(f"Top Student: {high} ({high_score})")

def lowest():
    student_table_sorted_ascending = student_table.sort_values(by='avg', ignore_index = True)
    low = student_table_sorted_ascending.loc[0,'name']
    low_score = student_table_sorted_ascending.loc[0,'avg']
    print(f"Lowest Student: {low} ({low_score})")

highest()
lowest()
```

```python
# Grade distribution
def grade_distribution():
    A = 0
    B = 0
    C = 0
    D = 0
    F = 0
    for i in student_table['grade']:
        if i == 'A':
            A += 1
        elif i == 'B':
            B += 1
        elif i == 'C':
            C += 1
        elif i == 'D':
            D += 1
        else:
            F += 1
    print(f"Grade Distribution: A-{A}, B-{B}, C{C}, D{D}, F{F}")

grade_distribution()
```
